```python
import os
import shutil

# --- KONFIGURASI ---
source_folder = r'R:\Rafi Ridho Ramadhan\KULIAH\SKRIPSI\Program Skripsi\Bachelor-Thesis-v1.1.1\data\processed\lampung - Copy'          
destination_folder = r'E:\lampung' 

os.makedirs(destination_folder, exist_ok=True)
```

```python
# 1. Ambil daftar file .npz
print("Membaca daftar file...")
files = [f for f in os.listdir(source_folder) if f.endswith('.npz')]
files.sort()

# 2. Hitung jumlah (setengah dari 42.947 adalah 21.473)
total_files = len(files)
copy_count = total_files // 2 

print(f"Total: {total_files} file. Akan disalin: {copy_count} file.")
print("-" * 30)
```

```python
# 3. Proses Copy
for i in range(copy_count):
    filename = files[i]
    src = os.path.join(source_folder, filename)
    dst = os.path.join(destination_folder, filename)
    
    try:
        # Menggunakan copy2 agar metadata tetap terjaga
        shutil.copy2(src, dst)
        
        if (i + 1) % 500 == 0:
            print(f"Progress: {i + 1}/{copy_count} file tersalin...")
            
    except Exception as e:
        print(f"Gagal menyalin {filename}: {e}")

print("-" * 30)
print("PROSES COPY SELESAI!")
```

```python
# Kita mengambil file dari index 'half_count' sampai akhir (:)
half_count = total_files // 2
files_part_2 = files[half_count:] 

count_part_2 = len(files_part_2)

print(f"Total File Asli: {total_files}")
print(f"File yang sudah dipindah (Tahap 1): {half_count}")
print(f"File yang akan disalin SEKARANG (Tahap 2): {count_part_2}")
print("-" * 30)
```

```python
# 3. Eksekusi Copy
for i, filename in enumerate(files_part_2):
    src = os.path.join(source_folder, filename)
    dst = os.path.join(destination_folder, filename)
    
    try:
        shutil.copy2(src, dst)
        
        # Tampilkan progress
        if (i + 1) % 500 == 0:
            print(f"Progress: {i + 1}/{count_part_2} file tersalin...")
            
    except Exception as e:
        print(f"Gagal menyalin {filename}: {e}")

print("-" * 30)
print("SELESAI! Bagian kedua sudah diamankan.")
```
